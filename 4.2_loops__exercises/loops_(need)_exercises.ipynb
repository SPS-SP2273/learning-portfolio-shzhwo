```python
for i in range(1, 6):
    print(f'{i} :', end='\t')
    for j in range(1, 6):
        print(f'{i*j}', end='\t')
    print()
```

```python
import numpy as np

for i in range(10):
    np_temp_array = np.random.rand(10)
    np_mask = np_temp_array >= 0.5
    total_heads = np.sum(np_mask)
    print(f'Experiment {i+1:>2}: No. of Heads = {total_heads}')
```
