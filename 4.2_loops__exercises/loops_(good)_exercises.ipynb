# Exercise 1

```python
squares = []

for i in range(10):
    if i % 2 == 0:
        continue # continue inserted for the sake of following the instruction
    squares.append(i ** 2)

print(squares)
```

# Exercise 2

```python
squares_2 = [i**2 for i in range(10) if i % 2 == 1]

print(squares_2)
```

# Exercise 4

```python
radioactivity = 100
half_radioactivity = 0.5 * radioactivity
decay_rate = 0.05

time_years = 0

while radioactivity > half_radioactivity:
    radioactivity *= (1 - decay_rate)
    time_years += 1

print(time_years)
```

# Exercise 5

```python
co2_data = [
    (2000, 369.55), (2001, 371.14), (2002, 373.28), 
    (2003, 375.80), (2004, 377.52), (2005, 379.80), 
    (2006, 381.90), (2007, 383.79), (2008, 385.60), 
    (2009, 387.43), (2010, 389.90), (2011, 391.65), 
    (2012, 393.85), (2013, 396.52), (2014, 398.65),
    (2015, 400.83), (2016, 404.24), (2017, 406.55), 
    (2018, 408.52), (2019, 411.44), (2020, 414.24)
]

base_co2 = co2_data[0][1] # assuming non-empty list

co2_increase_3ppm_list = []

for year_co2 in co2_data:
    if year_co2[1] - 3 > base_co2:
        percent_increase = (year_co2[1] - base_co2) * 100 / base_co2
        co2_increase_3ppm_list.append((year_co2[0], percent_increase))
    base_co2 = year_co2[1]

print("The following list shows years which observed an increase of CO2 concentration of at least 3 ppm compared to the previous year with the corresponding percentage increase:")
print(co2_increase_3ppm_list)
```
