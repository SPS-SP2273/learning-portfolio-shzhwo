```python
%%timeit

squares = [i**2 for i in range(10)]
```

Taking "661 ns ± 34.5 ns per loop (mean ± std. dev. of 7 runs, 1,000,000 loops each)" as an example, 661 ns is the average time it takes to run the cell once and 34.5 ns is the standard deviation calculated from timing the code. A run represents a timing period, so 7 runs mean that 7 timings were taken. In each run, the cell was ran for 1,000,000 times (number of loops) and the timing of running the cell 1,000,000 times was recorded.
