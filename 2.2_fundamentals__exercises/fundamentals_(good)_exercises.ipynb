# Exercise 2

```python
student_grade = input('Enter the student\'s score: ')

def score_to_grade(raw_score: str) -> str:
    try:
        float(raw_score)
    except:
        return "Score does not appear to be numerical"
    
    score = int(float(raw_score)) # scores that are not integer will be assumed to be user error and rounded down to the nearest int

    match score:
        case score if score >= 70 and score <= 100:
            grade = 'A'
        case score if score >= 50 and score <= 69:
            grade = 'B'
        case score if score >= 35 and score <= 49:
            grade = 'C'
        case score if score >= 0 and score <= 34:
            grade = 'Fail'
        case _:
            return "Score appears to be outside of the specified range"
    
    return f"The student\'s letter grade is: {grade}"

score_to_grade(student_grade)
```

# Exercise 3

```python
def factorial(n):
    if n < 0:
        return "Invalid input"
    elif n == 0:
        return 1
    else:
        result = 1
        for number in range(1, n+1):
            result *= number
        return result


print(factorial(5))   # Testing, expected output: 120
print(factorial(-1))  # Testing, expected output: Invalid input
```
