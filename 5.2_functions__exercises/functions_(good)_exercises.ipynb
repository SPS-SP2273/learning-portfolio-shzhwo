# Exercise 1

```python
def convert_celsius(temperature_celsius: float, target_scale: str='fahrenheit') -> float:
    lowercase_target_scale = target_scale.lower()

    match lowercase_target_scale:
        case "fahrenheit":
            return temperature_celsius * 1.8 + 32
        case "kelvin":
            return temperature_celsius + 273.15
        case _:
            print("The scale entered is not recognised by this function.")
            return
    
    # placeholder code
    print("end of function")
    return

print(convert_celsius(50, "Fahrenheit"))
print(convert_celsius(50, "FahrenHEIT"))
print(convert_celsius(50, "KELVIN"))
print(convert_celsius(50, "kelvin"))
print(convert_celsius(50, "rubbish"))
print(convert_celsius(50))
```

# Exercise 2

```python
def convert_fahrenheit(temperature_fahrenheit: float, target_scale: str='Celsius'):
    lowercase_target_scale = target_scale.lower()

    match lowercase_target_scale:
        case "celsius":
            return (temperature_fahrenheit - 32) / 1.8
        case "kelvin":
            return ((temperature_fahrenheit - 32) / 1.8) + 273.15
        case _:
            print("The scale entered is not recognised by this function.")
            return

    # placeholder code
    print("end of function")
    return

print(convert_celsius(95, "CELSIUS"))
print(convert_celsius(95, "celSIUS"))
print(convert_celsius(95, "Kelvin"))
print(convert_celsius(95, "kelvin"))
print(convert_celsius(95, "rubbish"))
print(convert_celsius(95))
```

# Exercise 3

```python
def convert_temperature(temperature: float, source_scale: str, target_scale: str) -> float:
    lowercase_source_scale = source_scale.lower()
    lowercase_target_scale = target_scale.lower()

    scale_states = ["celsius", "fahrenheit", "kelvin"]

    if lowercase_source_scale not in scale_states:
        print("The source scale entered is not recognised by this function.")
        return

    if lowercase_target_scale not in scale_states:
        print("The target scale entered is not recognised by this function.")
        return

    temperature_celsius = 0

    if lowercase_source_scale == "kelvin":
        temperature_celsius = temperature - 273.15
    
    elif lowercase_source_scale == "fahrenheit":
        temperature_celsius = convert_fahrenheit(temperature, lowercase_target_scale)
    
    else:
        temperature_celsius = temperature
    
    return convert_celsius(temperature_celsius, lowercase_target_scale)
```
