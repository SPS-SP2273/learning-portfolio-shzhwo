# Exercise 1

```python
def greeting(name):
    if name == 'Batman':
        return 'Hello Batman! So, nice to meet you!'
    return f'Hello {name}!'
```

The function above works as intended because returning a string in the if block prevents the rest of the function from being executed.

# Exercise 2

```python
def calculate_bmi(weight: float, height: float) -> str:
    if weight < 0 or height < 0:
        return 'This function does not accept negative values for weight or height parameters.'

    bmi = weight / (height ** 2)
    bmi_rounded = round(bmi, 1)

    classification = ""

    match bmi:
        case bmi if bmi >= 30:
            classification = "obese"
        case bmi if bmi >= 25 and bmi < 30:
            classification = "overweight"
        case bmi if bmi >= 18.5 and bmi < 24.9:
            classification = "normal weight"
        case bmi if bmi < 18.5:
            classification = "underweight"
    
    return f"Your BMI classifies you to be {classification}"

print(calculate_bmi(45, 1.5))
```

# Exercise 3

```python
import math

def factorial(n: int) -> float:
    if n < 0:
        print("This function does not accept negative values as input.")
        return 
    return math.gamma(n+1)

print(factorial(-1))
print(factorial(0))
print(factorial(1))
print(factorial(5))
```
