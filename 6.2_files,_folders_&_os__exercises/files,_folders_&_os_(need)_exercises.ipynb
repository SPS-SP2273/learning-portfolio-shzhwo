```python
# Task 1

table = {
'jan': '01', 'feb': '02', 'mar': '03',
'apr': '04', 'may': '05', 'jun': '06',
'jul': '07', 'aug': '08', 'sep': '09',
'oct': '10', 'nov': '11', 'dec': '12'
}

raw_filename = 'oct-08_singapore.txt'

first_split = raw_filename.split('-')
second_split = first_split[1].split('_')

month = table[first_split[0]]
day = second_split[0]
city = second_split[1]

processed_filename = '-'.join([month, day, city])

print(processed_filename)
```

```python
# Task 2

table = {
'jan': '01', 'feb': '02', 'mar': '03',
'apr': '04', 'may': '05', 'jun': '06',
'jul': '07', 'aug': '08', 'sep': '09',
'oct': '10', 'nov': '11', 'dec': '12'
}

def rename_my_file(old_file_name: str) -> str:

    raw_filename = old_file_name

    first_split = raw_filename.split('-')
    second_split = first_split[1].split('_')

    month = table[first_split[0]]
    day = second_split[0]
    city = second_split[1]

    processed_filename = '-'.join([month, day, city])

    return processed_filename

print(rename_my_file('oct-08_singapore.txt'))
```

```python
# Task 3

files_to_rename = ['oct-08_singapore.txt', 'jul-10_cairo.txt', 'may-15_dhaka.txt',
 'may-13_cairo.txt', 'oct-21_cairo.txt', 'jan-10_singapore.txt',
 'jun-20_tokyo.txt', 'aug-06_jakarta.txt', 'dec-21_karachi.txt',
 'jan-01_tokyo.txt']

processed_files = []

for file in files_to_rename:
    processed_file = rename_my_file(file)
    print(f'{file} ----> {processed_file}')
    processed_files.append(processed_file)
```

```python
# Task 4

import os
import glob

folder = 'os-collaboration-exercise-data'

file_list = glob.glob(f'{folder}/*')

for filepath in file_list:
    try:
        filename = os.path.split(filepath)[-1]
        renamed_file = rename_my_file(filename)
        new_filepath = os.path.join(folder, renamed_file)
        os.rename(filepath, new_filepath)
    except:
        print(f"Something went wrong renaming the file {file}")
```

```python
# Task 5

import os

os.mkdir('cities')

cities = ['Cairo', 'Dhaka', 'Jakarta', 'Karachi', 'Manila',
 'Melbourne', 'Osaka', 'Shanghai', 'Singapore', 'Tokyo']

for city in cities:
    try:
        city_path = os.path.join('cities', city)
        print(f'Creating {city_path}')
        os.mkdir(city_path)
    except FileExistsError:
        print(f'{city_path} already exists, skipping creation')
```

```python
# Task 6

import os
import glob
import shutil

city_path_cairo = os.path.join('cities', 'cairo')
city_files_cairo = glob.glob(f'os-collaboration-exercise-data/*cairo.*')

for city_file_cairo in city_files_cairo:
        shutil.copy(city_file_cairo, city_path_cairo)



# Task 7

cities_without_cairo = ['Cairo', 'Dhaka', 'Jakarta', 'Karachi', 'Manila',
 'Melbourne', 'Osaka', 'Shanghai', 'Singapore', 'Tokyo']

for city in cities_without_cairo:
    city_path = os.path.join('cities', city)
    city_files = glob.glob(f'os-collaboration-exercise-data/*{city}.*')
    
    for city_file in city_files:
        shutil.copy(city_file, city_path)
```
