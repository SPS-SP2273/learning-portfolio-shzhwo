#### Two similarities
1. Ordering of elements is preserved in both lists and NumPy arrays
2. _Elements of lists and NumPy arrays can be accessed and modified directly by indexing_

#### Two differences
1. NumPy arrays support matrix and vector operations for numeric arrays, lists do not have such a parallel
2. All elements in a NumPy array are of the same data type, there is no such restriction on lists

#### Dictionary
A dictionary is a function of an unordered list of objects such that the value of the function applied to these objects are accessed from memory rather than by computation.

# Exercise 5

```python
import numpy as np

x_1 = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
x_2 = [1, 8, 27, 64, 125, 216, 343, 512, 729, 1000]

x_1_np = np.array(x_1)
x_2_np = np.array(x_2)

# mean
x_1_mean = np.mean(x_1_np)
x_2_mean = np.mean(x_2_np)

print(f'x_1 mean: {x_1_mean}')
print(f'x_2 mean: {x_2_mean}')

# stdev
x_1_stdev = np.std(x_1_np)
x_2_stdev = np.std(x_2_np)

print(f'x_1 standard deviation: {x_1_stdev}')
print(f'x_2 standard deviation: {x_2_stdev}')

# some variables
n_1 = len(x_1)
n_2 = len(x_2)

x_1_sum = np.sum(x_1_np)
x_2_sum = np.sum(x_2_np)

x_1_residual = x_1_np - x_1_mean
x_2_residual = x_2_np - x_2_mean

x_1_residual_squared_sum = np.sum(x_1_residual * x_1_residual)
x_2_residual_squared_sum = np.sum(x_2_residual * x_2_residual)

# Pearson correlation calculated manually
sum_residual_crossmultiplied = np.sum(x_1_residual * x_2_residual)
pearson_r = sum_residual_crossmultiplied / (x_1_residual_squared_sum * x_2_residual_squared_sum)**0.5

print(f'Pearson correlation calculated manually: {pearson_r}')
print(f'Pearson correlation calculated with np.corrcoef: {np.corrcoef(x_1_np, x_2_np)}')
```

`np.corrcoef` gives the same value as the manually calculated correlation but returns a matrix instead of a single number. This suggests that entering $n$ vectors will return an $n \times n$ matrix $A$ whose entries $A_{ij} = A_{ji}$ which is the correlation between vector $i$ and vector $j$.
